# Predicting Content Refresh Opportunities Using Machine Learning and Search Intelligence

## Abstract

This study investigates whether machine learning can help identify content pages that should be prioritized for content refresh using historical search intelligence signals. A rule-based baseline and a Random Forest classifier were developed and compared using the same validation strategy. Historical features such as impressions, clicks, CTR, search volume, and content age were used while excluding future or label-derived information to avoid data leakage. The machine learning model captured more complex relationships than the baseline while the baseline remained useful because of its transparency and interpretability. The results are intended for decision-support and should be combined with human review before any production content changes.

## 1. Question

*The research question and the decision it supports.*

# Research Question

Can machine learning be used to identify content pages that should be prioritized for content refresh based on historical search performance and content quality signals?

This research supports the decision of which content pages should be reviewed first by SEO and content teams. The goal is to provide a ranked list of refresh candidates using measurable search intelligence signals. The output is intended for decision-support and does not replace human review.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

# Dataset

This study uses the FlyRank ML Internship dataset.

Dataset Release:
- FlyRank Internship Starter Dataset
- Approximately 30,000 anonymized content records

Each row represents one content item with historical search performance metrics.

Features used include:
- impressions_90d
- clicks_90d
- ctr
- search_volume
- word_count
- content_age_days

Excluded fields:
- trend_direction
- trend_pct
- is_declining_label (except when used as the prediction target)

These fields were excluded from feature engineering because they contain future information or label-derived information that could introduce data leakage.

The analysis uses anonymized public-safe data only. No client names, URLs, or private search queries are included.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

# Methodology

The objective is to prioritize content pages that are likely to benefit from a content refresh.

Feature selection focused on historical search performance and content characteristics available before prediction.

Features:
- Impressions
- Clicks
- CTR
- Search Volume
- Word Count
- Content Age

Baseline:
A rule-based scoring system combining visibility, freshness, and CTR.

Machine Learning Model:
Random Forest Classifier

Validation:
An 80/20 train-test split was used to compare the machine learning model with the baseline.

Leakage Prevention:
Label-derived variables such as trend_direction and trend_pct were excluded from the feature set. Only information available before prediction was used.
