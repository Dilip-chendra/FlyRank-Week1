## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

## Ranked Actions + Reason Codes

The ranked action queue prioritizes pages that may benefit from a content refresh based on observed search performance signals.

Pages with higher scores should be reviewed first because they combine strong visibility, older content, and relatively low click-through rates.

### Reason Codes

- **refresh_candidate** – High priority for content refresh.
- **stale_content** – Content is old and may require updating.
- **low_ctr** – Good visibility but low click-through rate.
- **high_visibility** – Receives many impressions and has improvement potential.

The recommendations are intended for decision-support and should always be reviewed by a content specialist before implementation.

```python
import os
import sys
import subprocess
import pandas as pd

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(
            ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
            check=True
        )
    os.chdir(REPO_DIR)

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["visibility_score"] = df["impressions_90d"] / df["impressions_90d"].max()
df["freshness_score"] = df["content_age_days"] / df["content_age_days"].max()
df["ctr_score"] = 1 - (df["ctr"] / df["ctr"].max())

df["action_score"] = (
    0.4 * df["visibility_score"] +
    0.4 * df["freshness_score"] +
    0.2 * df["ctr_score"]
)

df["reason_code"] = "refresh_candidate"
df["recommended_action"] = "Review and Refresh Content"

ranked = df.sort_values("action_score", ascending=False)

ranked[
    [
        "content_id",
        "action_score",
        "reason_code",
        "recommended_action"
    ]
].head(20)
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human Review

Before applying any recommendation, a reviewer should verify:

- Content quality
- Search intent
- Business importance
- Recent content updates
- Seasonal relevance

## No-Go List

The following should never be automated:

- Publishing content automatically
- Deleting pages automatically
- Updating legal or compliance content
- Refreshing recently updated pages
- Acting on low-confidence recommendations
