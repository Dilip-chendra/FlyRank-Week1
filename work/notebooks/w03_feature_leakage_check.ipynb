## 1. Build the feature vector

## Feature Vector

The feature vector contains historical search performance and content characteristics that are available before prediction.

Selected features:
- impressions_90d
- clicks_90d
- ctr
- search_volume
- word_count
- content_age_days

Missing numerical values are filled with 0 for this baseline analysis. These features are available before the prediction point and do not contain future information.

```python
import os
import sys
import subprocess
import pandas as pd

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if "google.colab" in sys.modules:
    if not os.path.exists(REPO_DIR):
        subprocess.run(
            ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
            check=True
        )
    os.chdir(REPO_DIR)

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

features = [
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "search_volume",
    "word_count",
    "content_age_days"
]

X = df[features].fillna(0)

print("Feature Vector Shape:", X.shape)

X.head()
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

## Leakage Check

The feature set was reviewed for potential leakage.

The following columns were intentionally excluded because they contain future information or are derived from the prediction label:

- trend_direction
- trend_pct
- is_declining_label

No label-derived or future-window features were included in the feature vector.

```python
excluded = [
    "trend_direction",
    "trend_pct",
    "is_declining_label"
]

print("Leakage Check")

for column in excluded:
    print("-", column)

print("\nNo excluded columns appear in the feature vector.")

print(set(excluded).intersection(set(X.columns)))
```
