## My Lane: Refresh / Content Opportunity Scoring

I chose the **Refresh / Content Opportunity Scoring** lane because SEO teams often have many webpages but limited time to review them. Instead of checking every page manually, I want to build a system that ranks pages based on their likelihood of needing a content refresh. This can help content teams prioritize their work and focus on pages with the greatest opportunity for improvement. The goal is to support human decision-making with data rather than replace it.

## The Question

### Research Question
Which webpages should be reviewed first for content refresh based on their search performance and content signals?

### Decision
The decision is which pages should be prioritized for review and refresh.

### Unit of Analysis
Each row represents one webpage.

### Output
A ranked list of webpages with a refresh priority score and recommendation.

### Action
SEO or content teams review and update the highest-ranked pages first.

### Cost of a Wrong Recommendation
If the model recommends the wrong pages, time and resources may be spent updating pages that do not need attention while important declining pages are missed. This could reduce the effectiveness of SEO efforts.

### Why Machine Learning?
Machine learning can combine multiple signals such as impressions, clicks, CTR, average position, content age, and engagement metrics to identify patterns that are difficult to capture with simple manual rules.

## Quick Look at the Data

The following code loads the starter dataset and shows a few statistics that support this project direction.

```python
import os
import sys
import subprocess
import json
import pandas as pd

# ==========================================================
# SETUP
# ==========================================================

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        print("Cloning repository...")
        subprocess.run(
            ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
            check=True
        )

    os.chdir(REPO_DIR)

    print("Installing requirements...")
    subprocess.run(
        [sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"],
        check=True
    )

else:
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

print("=" * 70)
print("Working Directory:", os.getcwd())
print("=" * 70)

assert os.path.exists(
    "data/raw/content_refresh_anonymized.csv"
), "Starter dataset not found."

print("Starter dataset found.")

# ==========================================================
# RUN PIPELINE
# ==========================================================

print("\nRunning ML Pipeline...\n")

subprocess.run(
    [sys.executable, "scripts/run_all.py"],
    check=True
)

# ==========================================================
# LOAD DATASET
# ==========================================================

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("\n" + "=" * 70)
print("DATASET SUMMARY")
print("=" * 70)

print("Total Pages   :", len(df))
print("Total Columns :", len(df.columns))

# ==========================================================
# REQUIRED NUMBERS
# ==========================================================

print("\nTrend Direction Counts")
print(df["trend_direction"].value_counts())

print("\nCorrelation: Search Volume vs Impressions")
corr = df["search_volume"].corr(df["impressions_90d"])
print(round(corr, 3))

print("\nMedian Word Count by Trend Direction")
print(df.groupby("trend_direction")["word_count"].median())

print("\nAverage CTR")
print(round(df["ctr"].mean(), 4))

print("\nAverage Position")
print(round(df["avg_position"].mean(), 2))

# ==========================================================
# MODEL RESULTS
# ==========================================================

results = json.load(open("outputs/model_results.json"))

baseline = results["baseline"]["baseline_precision_at_50"]
rf = results["models"]["random_forest"]["precision_at_50"]

print("\n" + "=" * 70)
print("MODEL PERFORMANCE")
print("=" * 70)

print(f"Baseline Precision@50      : {baseline:.3f}")
print(f"Random Forest Precision@50 : {rf:.3f}")
print(f"Improvement                : {rf/baseline:.2f}x")

print("\nValidation Strategy :", results["split_strategy"])

print("\nNotebook is ready for Week 1 Research Question.")
```

### Observations

- The starter dataset contains approximately **30,000 webpages**.
- The correlation between search volume and impressions is close to **0**, indicating that search volume alone is not a strong predictor of page impressions.
- Pages with different trend directions have similar median word counts, suggesting that content length alone is unlikely to explain performance differences.
