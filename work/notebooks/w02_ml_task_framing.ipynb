## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My Lane as an ML Task

I selected **Refresh / Content Opportunity Scoring**.

This project is mainly a **classification and scoring** task. The model predicts whether a webpage is likely to decline in performance and assigns a score so that SEO teams can prioritize which pages should be reviewed first. Instead of reviewing every page manually, the model ranks pages based on their likelihood of needing a content refresh. This helps teams save time and focus on the most important pages.

## Target or Proxy

The target used in the starter dataset is **is_declining_label**.

This label is created from the observed column **trend_direction**. If the trend direction is **down**, the page is labeled as declining; otherwise it is not.

This is a proxy label based on observed data. It helps train the model to identify pages that may require a content refresh.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

The primary success metric for this project is **Precision@50**.

Precision@50 measures how many of the top 50 pages recommended by the model are actually declining pages.

A higher Precision@50 means the review team spends their time on pages that are more likely to need attention. This metric matches the real business goal because teams usually review only a limited number of pages.

## Unit of Analysis

In this dataset, **one row represents one webpage (content page)**.

Each row contains information about that page, including impressions, clicks, sessions, CTR, average position, content age, word count, and trend direction.

These features are used to estimate whether the page should be prioritized for a content refresh.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A fixed rule cannot capture all the factors that influence page performance.

A webpage may have high impressions but low CTR, or good engagement but declining traffic. These patterns are affected by many features working together.

Machine learning can learn relationships between multiple variables such as impressions, CTR, sessions, average position, content age, and engagement. This allows the model to rank pages more accurately than simple if-else rules.

The model is intended to support SEO decisions by prioritizing pages for review rather than replacing human judgment.
